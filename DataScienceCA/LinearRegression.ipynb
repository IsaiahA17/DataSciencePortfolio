Introduction
In this notebook, a linear regression will be performed on the columns of Ladder Score and Log GDP per capita for the 2024 World Happiness Report.
The Ladder Score is measured from 1 - 10 and represents the average satisfaction with life for a country, while the Log GDP column simply represents a country's log GDP.

```python
#Student Name: Isaiah Andres
#Student Number: C00286361
#Importing Dependencies

%matplotlib inline 
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-whitegrid')
import numpy as np 
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
```

In this cell, the .csv for the World Happiness Report will be read, and drop any countries that have missing data. The model is then trained on the  Log GDP(As a 2D Array) and the happiness columns and then makes a prediction on them.

```python
df = pd.read_csv('World-happiness-report-2024.csv').dropna()

logGDP = df['Log GDP per capita'].values
happiness = df['Ladder score'].values

X_train, X_test, y_train, y_test = train_test_split(
    logGDP[:, np.newaxis], happiness, test_size=0.2, random_state=42
)

model = LinearRegression(fit_intercept=True)
model.fit(X_train, y_train)

# Predictions
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

# Fit line for plotting
xfit = np.linspace(logGDP.min(), logGDP.max())
yfit = model.predict(xfit[:, np.newaxis])

# Plot
plt.scatter(logGDP, happiness, label="Data")
plt.plot(xfit, yfit, label="Fitted Line", color="red")

plt.xlabel('Log GDP')
plt.ylabel('Life Satisfaction 0 - 8')
plt.title('Satisfaction With Life In Countries by Log GDP')
plt.legend()
plt.show()

# Model parameters
print("Model slope:    ", model.coef_[0])
print("Model intercept:", model.intercept_)

# R-squared scores
r2_train = r2_score(y_train, y_train_pred)
r2_test = r2_score(y_test, y_test_pred)

print(f"R-squared on training data: {r2_train}")
print(f"R-squared on test data: {r2_test}")
```

After making predictions with the model, the matplotlib library will be used to add a title and plot the points and draw the line to represent the linear regression. The r2 score may represent how well the line fits the model and the slope shows how much the ladder score goes up per 0.5 of the Log GDP, and the intercept shows what the ladder score may be when at 0. In conclusion, this model predicts that the line of best fit appears to be positive where the happiness or satisfaction with life goes up with as the Log GDP of a country increases.
